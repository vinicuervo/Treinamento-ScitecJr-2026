```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('ai_student_impact_dataset_final.csv')
```

```python
sns.boxplot(
    data=df,
    x='Major_Category',
    y='Perceived_AI_Dependency'
)

plt.xticks(
    ticks=[0,1,2,3,4],
    labels=['Artes', 'Negócios', 'Humanas', 'Medicina', 'STEM']
)

plt.title('Dependência de IA por Área de Formação')
plt.xlabel('Área de Formação')
plt.ylabel('Nível de Dependência')

plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Anxiety_Level_During_Exams',
    y='Weekly_GenAI_Hours'
)

plt.xlabel('Horas Semanais de IA')
plt.ylabel('Nível de Ansiedade em Provas')
plt.title('Relação entre uso de IA e Ansiedade em Provas')

plt.show()
```

```python
sns.boxplot(
    data = df,
    x = 'Burnout_Risk_Level',
    y = 'Weekly_GenAI_Hours'
)

plt.xlabel('Risco de Burnout')
plt.ylabel('Horas Semanais de IA')
plt.title('Relação entre uso de IA e Burnout')

plt.show()
```

```python
corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f'
)

plt.title('Heatmap de Correlação')

plt.show()
```

```python
sns.histplot(
    data=df,
    x='Weekly_GenAI_Hours'
)

plt.xlabel('Horas Semanais de IA')
plt.ylabel('Quantidade de Estudantes')
plt.title('Distribuição de IA entre os estudantes')

plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Year_of_Study',
    y='Weekly_GenAI_Hours'
)

plt.xlabel('Ano de Estudo')
plt.ylabel('Horas Semanais de IA')
plt.title('Uso de IA por Ano de Estudo')

plt.show()
```

```python
df["Faixa"] = pd.cut(
    df["Traditional_Study_Hours"],
    bins=[0,5,10,15,20,25,30,35,40]
)

media = (
    df.groupby("Faixa", observed=True)["Weekly_GenAI_Hours"]
      .mean()
)

plt.figure(figsize=(9,5))

media.plot(marker='o')

plt.title("Uso médio de IA por faixa de estudo tradicional")
plt.xlabel("Horas de estudo tradicional")
plt.ylabel("Média de horas de IA")

plt.grid(True)

plt.show()
```

```python
agrupado = (
    df.groupby("Major_Category")[[
        "Traditional_Study_Hours",
        "Weekly_GenAI_Hours"
    ]]
    .mean()
)

agrupado.plot(
    kind="bar",
    stacked=True,
    figsize=(10,6)
)

plt.ylabel("Horas médias por semana")
plt.title("Horas médias de estudo por curso")

plt.show()
```

```python
df["Melhora_GPA"] = (
    df["Post_Semester_GPA"] -
    df["Pre_Semester_GPA"]
)

plt.scatter(
    df["Weekly_GenAI_Hours"],
    df["Melhora_GPA"],
    alpha=0.3
)

plt.xlabel("Horas semanais de IA")
plt.ylabel("Melhora no GPA")

plt.show()
```
