```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('ai_student_impact_dataset_final.csv')
```

```python
sns.boxplot(
    data=df,
    x='Major_Category',
    y='Perceived_AI_Dependency'
)

plt.xticks(
    ticks=[0,1,2,3,4],
    labels=['Artes', 'Negócios', 'Humanas', 'Medicina', 'STEM']
)

plt.title('Dependência de IA por Área de Formação')
plt.xlabel('Área de Formação')
plt.ylabel('Nível de Dependência')

plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Anxiety_Level_During_Exams',
    y='Weekly_GenAI_Hours'
)

plt.xlabel('Horas Semanais de IA')
plt.ylabel('Nível de Ansiedade em Provas')
plt.title('Relação entre uso de IA e Ansiedade em Provas')

plt.show()
```

```python
sns.boxplot(
    data = df,
    x = 'Burnout_Risk_Level',
    y = 'Weekly_GenAI_Hours'
)

plt.xlabel('Risco de Burnout')
plt.ylabel('Horas Semanais de IA')
plt.title('Relação entre uso de IA e Burnout')

plt.show()
```

```python
sns.regplot(
    data=df,
    x='Weekly_GenAI_Hours',
    y='Post_Semester_GPA',
    scatter_kws={'alpha':0.1}
)

plt.xlabel('Horas Semanais de IA')
plt.ylabel('Pós-Semestre GPA')
plt.title('Relação entre horas de IA e GPA')

plt.show()
```

```python
corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f'
)

plt.title('Heatmap de Correlação')

plt.show()
```

```python
sns.histplot(
    data=df,
    x='Weekly_GenAI_Hours'
)

plt.xlabel('Horas Semanais de IA')
plt.ylabel('Quantidade de Estudantes')
plt.title('Distribuição de IA entre os estudantes')

plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Year_of_Study',
    y='Weekly_GenAI_Hours'
)

plt.xlabel('Ano de Estudo')
plt.ylabel('Horas Semanais de IA')
plt.title('Uso de IA por Ano de Estudo')

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["Traditional_Study_Hours"],
    df["Weekly_GenAI_Hours"],
    alpha=0.15,
    s=8     
)

plt.title("Horas de Estudo Tradicional x Uso de IA")
plt.xlabel("Horas de Estudo Tradicional")
plt.ylabel("Horas de Uso de IA")

plt.show()
```

```python
df["Total_Study_Hours"] = (
    df["Traditional_Study_Hours"] +
    df["Weekly_GenAI_Hours"]
)

print(df[["Traditional_Study_Hours",
          "Weekly_GenAI_Hours",
          "Total_Study_Hours"]].head())

medias = [
    df["Traditional_Study_Hours"].mean(),
    df["Weekly_GenAI_Hours"].mean()
]

nomes = [
    "Estudo Tradicional",
    "Uso de IA"
]

plt.figure(figsize=(6,5))
plt.bar(nomes, medias)

plt.title("Média de Horas Semanais")
plt.ylabel("Horas")

plt.show()
```

```python
colunas = [
    "Traditional_Study_Hours",
    "Weekly_GenAI_Hours",
    "Pre_Semester_GPA",
    "Post_Semester_GPA",
    "Skill_Retention_Score"
]

corr = df[colunas].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlação entre Variáveis")

plt.show()
```
